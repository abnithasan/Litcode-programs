```python
#phone mapping
phone={'2':['a','b','c'],'3':['d','e','f'],'4':['g','h','i'],'5':['j','k','l'],'6':['m','n','o'],'7':['p','q','r','s'],'8':['t','u','v'],'9':['w','x','y','z']}
n=input()
l=['']

for i in range(len(n)):
    if n[i] in phone:
        alpha=phone[n[i]]

    l1=[]
    for j in l:
        for k in alpha:
            l1.append(j+k)
    l=l1

print(l)
```

```python
#clumsy factorial
num=list(map(int,input().split()))
l=[]
if len(num)>1:
    for i in range(num[0],num[1]-1,-1):
        if i==0:
            break
        else:
            l.append(i)
else:
    for i in range(num[0],0,-1):
        if i==0:
            break
        else:
            l.append(i)
odd=0;even=0
s=str(l[0])
for i in range(1,len(l)):
    if l[i-1]%2==0:
        if even==0:
            s+="//"+str(l[i])
            even=1
        else:
            s+="*"+str(l[i])
            even=0
    else:
        if odd==0:
            s+="+"+str(l[i])
            odd=1
        else:
            s+="-"+str(l[i])
            odd=0
print(eval(s))
```
