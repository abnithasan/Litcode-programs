```python

def FindAutoCount(n):
    d={}
    for i in n:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    flag=0
    for i in range(len(n)):
        if int(n[i])!=d.get(str(i),0):
            flag=1
            break

    if flag==1:
        print(0)
    else:
        print(len(d)
        )

n=input()
FindAutoCount(n)

```

```python
#Toggle all bits
n=int(input())
n=bin(n)[2:]
new=""

for i in n:
    if i=="0":
        new+=str(1)
    if i=="1":
        new+=str(0)

decimal=int(new,2)

print(decimal)


```

```python
#Corona Spikes
n=int(input())
number=list(map(int,input().split()))
s=int(input())
l=[]

for num in number:
    num=bin(num)[2:]
    d=num[:(len(num)-s)]
    l.append(d)

for i in l:
    if i=="":
        print(0,end=" ")
    else:
        print(int(i,2),end=" ")
```
